```python
#Series (index value is default)
import pandas as pd
data=[1,2,3,4,5]
s=pd.Series(data)
print(s)
```

```python
#Series (changing index value)
data=[1,2,3,4,5]
s=pd.Series(data,index=['A','B','C','D','E'])
print(s)
```

```python
#Series (Dictionary)
data={'name':'Rihana','age':20,'place':'india'}
s=pd.Series(data)
print(s)
```

```python
#loc used for accesing elements from customized index
nums=[25,35,45,55,65]
a=pd.Series(nums,index=['a','b','c','d','e'])
print(a.loc['a'])
print(a.loc['a':'c']) #will include last element in slicing
```

```python
#iloc is integer based accessing ie, for default index
nums=[25,35,45,55,65]
a=pd.Series(nums)
print(a.iloc[1])
print(a.iloc[:2:1]) #will exclude last element in slicing
```

```python
import pandas as pd
sales=pd.Series(
    [1200,850,1500,950,2000],
    index=['mon','tue','wed','thurs','fri']
)
print(sales)
print(sales.max())
print(sales.min())
print(sales.sum())
print(sales[sales<1000])
```

```python
import pandas as pd
data=[20,20,30,30,40,50,60,60,80]
ar=pd.Series(data)
print(ar)
print(ar.unique())  #to remove duplicate values
print(ar.nunique()) #to count the unique numbers
print(ar.value_counts()) #to show how many times a number repeated
```

```python
import pandas as pd
data=[20,None,30,30,None,50,None,60,80]
ar=pd.Series(data)
print(ar.isna()) #show TRUE for null values
print(ar.notna()) #show FALSE for null values
print(ar.dropna()) #prints not null values
```
