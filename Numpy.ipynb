```python
import numpy as np
mark=np.array([78,92,65,88,45,90,72,55])
print(mark)
print(np.sum(mark))
print(np.average(mark))
print(np.max(mark))
print(np.min(mark))
```

```python
#marks greater than 75
mark=np.array([78,92,65,88,45,90,72,55])
print(mark[mark>75])
#marks lesser than 50
print(mark[mark<=50])
#sorting
print(np.sort(mark))
```

```python
list=np.arange(1,21)
print(list)
```

```python
#zeros
r=np.zeros(5)
print(r)
s=np.ones(4)
print(s)
```

```python
q=np.random.randint(1,100,10)
print(q)
print(np.shape(q))
```

```python
temp=np.array([32,35,40,28,42,33])
print(np.mean(temp))
print(np.max(temp))
print(np.min(temp))

temp[temp>38]=38
print(temp)

t=temp.reshape(2,3)
print(t)

print(t[0])
print(t[:,1])
```

```python
#slicing
a=[10,20,30,40,50,60,70,80]
print(a[0::7])
print(a[2:6:1])
print(a[::2])
```

```python
a=np.array([10,20,30,40,50,60,79,80])
print(np.where(a%2==0))
print(np.where(a%2!=0))
print(a[a%2==0])
print(a[a%2!=0])
```
